# Group 3: Hours Worked vs. Earnings (Seaborn)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Create a dummy DataFrame for demonstration
np.random.seed(42)
num_samples = 200

data = {
    'hours worked per week': np.random.randint(20, 60, num_samples),
    'weekly earnings': np.random.normal(loc=800, scale=200, size=num_samples),
    'sex': np.random.choice(['Male', 'Female'], num_samples)
}
df = pd.DataFrame(data)

# Introduce some variation based on sex and hours worked
df.loc[df['sex'] == 'Female', 'weekly earnings'] = df.loc[df['sex'] == 'Female', 'weekly earnings'] * 0.9 + np.random.normal(loc=50, scale=100, size=(df['sex'] == 'Female').sum())
df.loc[df['sex'] == 'Male', 'weekly earnings'] = df.loc[df['sex'] == 'Male', 'weekly earnings'] * 1.1 + np.random.normal(loc=100, scale=150, size=(df['sex'] == 'Male').sum())

df['weekly earnings'] = df['weekly earnings'] + df['hours worked per week'] * 15 # Add a positive correlation
df['weekly earnings'] = df['weekly earnings'].apply(lambda x: max(100, x)) # Ensure earnings are not too low

display(df.head())
```

```python
# Create the scatter plot with trend lines, colored by sex
plt.figure(figsize=(10, 7))
sns.lmplot(
    data=df,
    x='hours worked per week',
    y='weekly earnings',
    hue='sex',
    scatter_kws={'alpha': 0.6},
    height=6, aspect=1.3,
    palette={'Male': 'skyblue', 'Female': 'lightcoral'}
)

plt.title('Weekly Earnings vs. Hours Worked per Week, by Sex')
plt.xlabel('Hours Worked per Week')
plt.ylabel('Weekly Earnings ($)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Analysis of Weekly Earnings vs. Hours Worked per Week

**Overall Relationship and Trend Lines:**
The scatter plot shows a general positive trend between 'hours worked per week' and 'weekly earnings', indicating that as hours worked increase, weekly earnings tend to increase as well. The trend lines (regression lines) for both 'Male' and 'Female' capture this overall upward slope. The overlap between the two trend lines suggests a similar general relationship for both sexes, but the vertical separation (if any) could indicate differences in average earnings for a given number of hours.

**Differences by Sex:**
By coloring points by sex, we can observe if this relationship differs. Visually, if one trend line is consistently higher than the other, it would suggest a disparity in earnings for the same amount of hours worked. For instance, if the 'Male' trend line is higher, it implies males, on average, earn more than females for the same hours. Conversely, if the 'Female' line is higher, the opposite is true. The spread of points around each trend line can also hint at differences in earning consistency or opportunity between sexes.

**Variability and Potential Outliers:**
- **Variability:** The spread of points around each trend line indicates the variability in earnings. A wider spread suggests more diverse earnings for a given number of hours worked, while a tighter cluster implies more consistent earnings. This variability can be observed for both sexes. For example, some individuals might work fewer hours but earn significantly more (or vice-versa), contributing to the scatter.
- **Potential Outliers:** Points that fall far away from the main cluster or the trend lines can be considered potential outliers. These could represent individuals with exceptionally high or low earnings for their hours worked, perhaps due to specialized skills, overtime pay, or unique employment situations. It's important to investigate these outliers further as they might reveal interesting insights or data anomalies.

**Interpretation of Trend Line (Descriptive vs. Causal):**
It is crucial to interpret the trend lines in this descriptive analysis carefully. While the lines show a statistical association, they **do not imply a causal relationship**. We cannot conclude that simply working more hours *causes* an increase in earnings due to this plot alone. There are many confounding factors that could influence both hours worked and earnings (e.g., job type, industry, experience, education, negotiation skills). This visualization merely describes the observed pattern in the data; it doesn't prove that increasing hours *directly leads* to higher earnings in a cause-and-effect manner.
